```
import pandas as pd
from google.colab import files

print("Por favor, selecione o arquivo 'uso_da_frota_2025.csv' no seu computador:")
uploaded = files.upload()

# 1. Carrega o arquivo da frota automaticamente após o upload
df = pd.read_csv("uso_da_frota_2025.csv", sep=";", encoding="latin1")
dados_passageiros = df['Passageiros'].dropna().tolist()

# 2. Média manual (necessária para a variância)
soma_total = 0
quantidade = 0
for val in dados_passageiros:
    soma_total += val
    quantidade += 1
media_manual = soma_total / quantidade

# 3. Mediana manual
def calcular_mediana(lista):
    lista_ordenada = sorted(lista)
    n = len(lista_ordenada)
    if n % 2 == 1:
        return lista_ordenada[n // 2]
    else:
        return (lista_ordenada[(n // 2) - 1] + lista_ordenada[n // 2]) / 2

# 4. Variância amostral manual
def calcular_variancia_amostral(lista, media):
    n = len(lista)
    if n <= 1:
        return 0
    soma_dos_quadrados = 0
    for val in lista:
        soma_dos_quadrados += (val - media) ** 2
    return soma_dos_quadrados / (n - 1)

# Executando e exibindo os resultados
mediana_val = calcular_mediana(dados_passageiros)
variancia_val = calcular_variancia_amostral(dados_passageiros, media_manual)
desvio_padrao_val = variancia_val ** 0.5

print("\n--- RESULTADOS DO NÚCLEO ESTATÍSTICO ---")
print(f"Média manual: {media_manual:.2f}")
print(f"Mediana manual: {mediana_val}")
print(f"Variância amostral manual: {variancia_val:.2f}")
print(f"Desvio padrão amostral manual: {desvio_padrao_val:.2f}")
```

```
import random
import matplotlib.pyplot as plt

# Módulo 3: Simulação de Monte Carlo - Lei dos Grandes Números
# Simulando lançamentos de um dado (exemplo clássico para mostrar convergência para 1/6 ≈ 16.67%)
def simular_lei_grandes_numeros(n_maximo):
    resultados = []
    sucessos = 0
    frequencias_relativas = []

    for i in range(1, n_maximo + 1):
        # Simula o lançamento de um dado de 6 lados (considerando '6' como o evento de sucesso)
        lancamento = random.randint(1, 6)
        if lancamento == 6:
            sucessos += 1
        freq_relativa = sucessos / i
        frequencias_relativas.append(freq_relativa)

    return frequencias_relativas

# Executando a simulação com 5.000 repetições
n_simulacoes = 5000
dados_simulacao = simular_lei_grandes_numeros(n_simulacoes)

# Plotando o gráfico da Lei dos Grandes Números
plt.figure(figsize=(10, 4))
plt.plot(range(1, n_simulacoes + 1), dados_simulacao, color='purple', label='Frequência Relativa')
plt.axhline(y=1/6, color='red', linestyle='--', label='Probabilidade Teórica (1/6 ≈ 16.67%)')
plt.title('Simulação de Monte Carlo: Lei dos Grandes Números')
plt.xlabel('Número de Simulações (N)')
plt.ylabel('Frequência Relativa do Evento')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```
import random
import matplotlib.pyplot as plt

# Módulo 3: Teorema Central do Limite (TCL) com os dados reais
dados_populacao = df['Passageiros'].dropna().tolist()

# Simulando amostragens repetidas e calculando as médias amostrais
tamanho_amostra = 30  # Tamanho de cada amostra
n_replicacoes = 1000   # Número de amostras tiradas
medias_amostrais = []

for _ in range(n_replicacoes):
    # Sorteia uma amostra aleatória da população com reposição
    amostra = [random.choice(dados_populacao) for _ in range(tamanho_amostra)]
    # Calcula a média da amostra
    media_amostra = sum(amostra) / len(amostra)
    medias_amostrais.append(media_amostra)

# Plotando a distribuição das médias amostrais (Módulo 3 & 4 - Tendência à Normal)
plt.figure(figsize=(10, 4))
plt.hist(medias_amostrais, bins=25, color='forestgreen', edgecolor='black', density=True)
plt.title('Teorema Central do Limite: Distribuição das Médias Amostrais')
plt.xlabel('Média de Passageiros da Amostra')
plt.ylabel('Densidade de Frequência')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```
import matplotlib.pyplot as plt

# Módulo 5: Correlação e Regressão Linear Simples (Método dos Mínimos Quadrados Manual)
# Selecionando duas colunas numéricas válidas para correlacionar (ex: índice vs Passageiros, ou limpando nulos de duas colunas)
df_reg = df[['Passageiros']].dropna()
# Vamos criar uma lista de índices numéricos como X e os passageiros como Y
X = list(range(len(df_reg)))[:1000]  # Pegando os primeiros 1000 para ficar rápido e visual
Y = df_reg['Passageiros'].tolist()[:1000]

n = len(X)
media_x = sum(X) / n
media_y = sum(Y) / n

# Cálculo manual dos coeficientes da reta de regressão (Inclinação 'a' e Intercepto 'b')
numerador = 0
denominador = 0
for i in range(n):
    numerador += (X[i] - media_x) * (Y[i] - media_y)
    denominador += (X[i] - media_x) ** 2

a = numerador / denominador  # Inclinação (slope)
b = media_y - (a * media_x)    # Intercepto (intercept)

# Gerando os valores preditos Y_pred para traçar a reta
Y_pred = [a * x + b for x in X]

# Exibindo os resultados da regressão
print(f"Equação da Reta: Ŷ = {a:.4f} * X + {b:.4f}")
print("Alerta: Correlação não implica causalidade!")

# Plotando o Diagrama de Dispersão e a Reta de Regressão
plt.figure(figsize=(10, 5))
plt.scatter(X, Y, color='gray', alpha=0.5, label='Dados Reais (Dispersão)')
plt.plot(X, Y_pred, color='red', linewidth=2, label='Reta de Regressão (Mínimos Quadrados)')
plt.title('Módulo 5: Regressão Linear Simples')
plt.xlabel('Variável X (Índice da Viagem)')
plt.ylabel('Variável Y (Passageiros)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
